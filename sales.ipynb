```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
weekly_sales = pd.read_csv('./data/salesweekly.csv')
```

```python
# What are the total sales quantities for each drug category (ATC code)?
print(monthly_sales.sum(numeric_only=True))
```

```python
# Which individual drug brands have the highest total sales?
highest_sales = monthly_sales.sum(numeric_only=True).max()
print(highest_sales)

# N02BE has the highest sales with 62477.945
```

```python
monthly_sales.head()
```

```python
# Which three drugs have the highest sales in January 2015, July 2016, September 2017.

# Concert column datum to date type
monthly_sales['datum'] = pd.to_datetime(monthly_sales['datum'])


jan_sales = (monthly_sales[
    (monthly_sales['datum'].dt.year == 2015) & 
    (monthly_sales['datum'].dt.month == 1)
])

july_sales = (monthly_sales[
    (monthly_sales['datum'].dt.year == 2016) & 
    (monthly_sales['datum'].dt.month == 7)
])
sept_sales = (monthly_sales[
    (monthly_sales['datum'].dt.year == 2017) & 
    (monthly_sales['datum'].dt.month == 9)
])

print(jan_sales.sum(numeric_only=True).sort_values(ascending=False).head(3))
print(july_sales.sum(numeric_only=True).sort_values(ascending=False).head(3))
print(sept_sales.sum(numeric_only=True).sort_values(ascending=False).head(3))

# Drugs with the higest sales in January 2015:
    # N02BE: 1044.24
    # N05B: 463.00
    # M01AB: 177.25

# Drugs with the higest sales in July 2016:
    # N02BE: 652.362
    # N05B: 240.000
    # M01AB: 203.970

# Drugs with the higest sales in September 2017:
# N02BE: 863.75
# N05B: 223.00
# R03: 139.00
```

```python
# Which drug has sold the most often in 2017?

sales_2017 = daily_sales[(daily_sales['Year'] == 2017)]
print(sales_2017[['M01AB', 'M01AE', 'N02BA', 'N02BE',
        'N05B', 'N05C', 'R03', 'R06']].sum(numeric_only=True).sort_values(ascending=False).head(1))

# The N02BE drug has sold the most in 2017 with 9258.80
```

```python
# Which drug category has the highest average daily sales?
print(
    daily_sales[['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']].mean().
    sort_values(ascending=False).head(1))

# N02BE has the highest average daily sales 29.917095
```

```python
# Are respiratory drugs (R03) sold more during specific months?
r03_sales = daily_sales[['R03', 'Month']]
r03_sales.groupby('Month').sum()

# R03 are sold more in the first three months
```
